```python
print("Exact duplicate rows:", posts_raw.duplicated().sum())

posts_final = posts_raw.drop_duplicates().copy()

print("Remaining rows:", len(posts_final))
```

```python
print("Missing values:")
print(posts_final.isna().sum())

print("\nRepeated post IDs:", posts_final["post_id"].duplicated().sum())
```

```python
columns = ["likes", "shares", "comments"]

missing_before = posts_final[columns].isna().sum()

for column in columns:
    posts_final[column] = pd.to_numeric(posts_final[column], errors="coerce")

print("Additional missing values caused by conversion:")
print(posts_final[columns].isna().sum() - missing_before)
```

```python
print("Negative values:")
print((posts_final[columns] < 0).sum())
```

```python

# Some NULL markers have spaces or newlines hiding around them.
text_check = posts_final["text_content"].astype("string").str.strip()
hidden_missing = text_check.str.fullmatch(r"(?i)(?:null|nan|none|n/a)?", na=False)

# We are only counting here, not changing your data.
print("Extra missing-text markers:", hidden_missing.sum())
print("Current missing text:", posts_final["text_content"].isna().sum())
print("Working rows:", len(posts_final))
```

```python
# Triming whitespace and handle pandas' default 'nan' string
posts_final['text_content'] = posts_final['text_content'].astype(str).str.strip()
posts_final['text_content'] = posts_final['text_content'].replace('nan', np.nan)

# Normalize hidden missing markers
missing_markers = ['', 'null', 'N/A', 'NA', 'None', 'NULL']
posts_final['text_content'] = posts_final['text_content'].replace(missing_markers, np.nan)
print(f"Missing text count after normalization: {posts_final['text_content'].isna().sum()}")
```

```python

#Handle Invalid (Negative) Likes

posts_final['likes_original'] = posts_final['likes']
posts_final['is_invalid_likes'] = posts_final['likes'] < 0
posts_final.loc[posts_final['is_invalid_likes'], 'likes'] = np.nan
print(f"Missing likes after invalidating negatives: {posts_final['likes'].isna().sum()}")
```

```python

# Timestamp Parsing

def parse_mixed_dates(date_series):
    numeric_dates = pd.to_numeric(date_series, errors='coerce')

    # Identify unix timestamps (e.g., 10 digits -> > 100,000,000)
    unix_mask = numeric_dates.notna() & (numeric_dates > 1e8)
    parsed_unix = pd.to_datetime(numeric_dates[unix_mask], unit='s', errors='coerce')

    # Parse remaining strings (ISO or d-m-y)
    string_mask = ~unix_mask
    parsed_strings = pd.to_datetime(date_series[string_mask], dayfirst=True, format='mixed', errors='coerce')

    # Combine results
    result = pd.Series(index=date_series.index, dtype='datetime64[ns]')
    result.update(parsed_unix)
    result.update(parsed_strings)
    return result

posts_final['timestamp_parsed'] = parse_mixed_dates(posts_final['timestamp'])
print(f"Missing parsed timestamps: {posts_final['timestamp_parsed'].isna().sum()}")
```

```python

# Validation and Joining

print(f"Unique post IDs: {posts_final['post_id'].nunique()}")
print(f"Unique user IDs in Users table: {users['user_id'].nunique()}")

# Merge datasets
df = posts_final.merge(users, on='user_id', how='left')
print(f"Joined dataframe shape: {df.shape}")
```

```python

# Exploratory Data Analysis & Plotting

missing_counts = df.isnull().sum()
missing_percentages = (missing_counts / len(df)) * 100
missing_df = pd.DataFrame({'Missing Count': missing_counts, 'Percentage': missing_percentages})
missing_df = missing_df[missing_df['Missing Count'] > 0].sort_values(by='Missing Count', ascending=False)

plt.figure(figsize=(8, 5))
sns.barplot(x=missing_df.index, y=missing_df['Percentage'], palette='viridis')
plt.title("Missing/Invalid Data Percentages")
plt.ylabel("% Missing")
plt.xticks(rotation=45)
plt.tight_layout()
plt.savefig("missing_data.png")
plt.show()

# Plot 2: Numerical Distributions
plt.figure(figsize=(12, 4))
plt.subplot(1, 3, 1)
sns.histplot(df['likes'].dropna(), bins=30, kde=True)
plt.title("Distribution of Valid Likes")
plt.subplot(1, 3, 2)
sns.histplot(df['shares'].dropna(), bins=30, kde=True)
plt.title("Distribution of Shares")
plt.subplot(1, 3, 3)
sns.histplot(df['comments'].dropna(), bins=30, kde=True)
plt.title("Distribution of Comments")
plt.tight_layout()
plt.savefig("num_distributions.png")
plt.show()


# Plot 3: Platform Contributions
plt.figure(figsize=(8, 5))
platform_counts = df['platform'].fillna("Unknown").value_counts()
sns.barplot(x=platform_counts.index, y=platform_counts.values, palette='mako')
plt.title("Posts by Platform")
plt.ylabel("Post Count")
plt.tight_layout()
plt.show()



# Plot 4: Engagement by Platform
plt.figure(figsize=(10, 6))
sns.boxplot(data=df.fillna({'platform': 'Unknown'}), x='platform', y='likes', showfliers=False, palette='Set2')
plt.title("Typical Engagement (Likes) by Platform (Excluding Extremes)")
plt.tight_layout()
plt.show()
```

```python

#Statistical Summaries

corr_cols = ['likes', 'shares', 'comments', 'follower_count']
corr_matrix = df[corr_cols].corr(method='spearman')
print("\nSpearman Correlations:\n", corr_matrix)

print("\nDescriptive stats for likes by platform:")
print(df.groupby(df['platform'].fillna('Unknown'))['likes'].describe()[['count', 'mean', '50%']])

print("\nTop 5 users by post count:")
print(df['user_id'].value_counts().head(5))

# Export Cleaned Dataset

output_filename = "Cleaned_Dataset.csv"
df.to_csv(output_filename, index=False)
print(f"\nSuccessfully exported cleaned dataset to {output_filename}")
```
